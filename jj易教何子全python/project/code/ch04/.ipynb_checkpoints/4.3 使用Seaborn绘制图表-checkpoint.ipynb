### 4.3.2 Seaborn图表主题

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 加载数据集
tips = sns.load_dataset('tips')

# 设置为 darkgrid主题（默认主题）
sns.set_style("darkgrid")
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
plt.show()
```

```python

# 设置为 whitegrid主题
sns.set_style("whitegrid")
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
plt.show()
```

#### 4.3.3 柱状图

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
# 设置中文字体和解决负号显示问题
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
# 读取CSV文件

# 示例数据：不同股票的年度收益
data = pd.DataFrame({
    'Stock': ['AAPL', 'GOOG', 'AMZN', 'MSFT', 'TSLA'],
    'Yearly Return': [0.45, 0.30, 0.35, 0.25, 0.40]
})

# 绘制柱状图
plt.figure(figsize=(10, 6))
sns.barplot(x='Stock', y='Yearly Return', data=data, hue='Stock',legend=False)

# 设置标题和标签
plt.title('股票年度收益比较', fontsize=16)
plt.xlabel('股票', fontsize=12)
plt.ylabel('年度收益率', fontsize=12)

# 显示图表
plt.show()
```

#### 4.3.4 直方图

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 假设我们有一只股票的历史收盘价数据
data = {
    'Date': ['2025-01-01', '2025-01-02', '2025-01-03', '2025-01-04', '2025-01-05', 
             '2025-01-06', '2025-01-07', '2025-01-08', '2025-01-09', '2025-01-10'],
    'Close': [120, 122, 121, 123, 125, 127, 129, 130, 132, 134]
}

# 将数据转化为DataFrame
df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])

# 绘制收盘价格的直方图
sns.histplot(df['Close'], kde=True, color='green', bins=5)

# 设置标题和标签
plt.title('股票历史收盘价分布', fontsize=16)
plt.xlabel('收盘价 (美元)', fontsize=12)
plt.ylabel('频数', fontsize=12)

# 显示图表
plt.show()
```

#### 4.3.6 小提琴图

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 设置中文显示
plt.rcParams['font.family'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
# 读取数据
data = pd.read_csv('data/stock_data.csv')

# 绘制小提琴图
sns.violinplot(x='Day', y='Price', data=data)

# 设置图表标题和轴标签
plt.title('股票收盘价')
plt.xlabel('日期')
plt.ylabel('收盘价')
```

#### 4.3.7 热力图

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 设置中文显示
plt.rcParams['font.family'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建测试数据
data = pd.DataFrame({
    '股票1': [10, 12, 8, 15, 9],
    '股票2': [20, 18, 25, 22, 24],
    '股票3': [7, 9, 6, 8, 10],
    '股票4': [13, 11, 14, 10, 12]
})

# 提取多个股票的收盘价数据列
股票_prices = data[['股票1', '股票2', '股票3', '股票4']]

# 计算收盘价之间的相关性
correlation_matrix = 股票_prices.corr()	

# 绘制热力图
plt.figure(figsize=(10, 8))  # 设置图形大小
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

# 设置图形标题
plt.title('股票 Prices Correlation')

# 显示图形
plt.show()
```
