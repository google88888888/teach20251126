### 10.3 封装海龟策略函数

```python
import numpy as np
import pandas as pd

# 定义海龟策略函数
def turtle_trading_strategy(df, ma_short_window=20, ma_long_window=50, initial_capital=1000000, unit_size=100, risk_per_trade=0.01, atr_window=14):
    """
    封装海龟交易策略：通过突破信号和风险管理进行买入和卖出决策，并基于风险参数调整仓位。
    
    :param df: 包含 'Close' 列的股票历史数据
    :param ma_short_window: 短期移动平均线的窗口期（默认20）
    :param ma_long_window: 长期移动平均线的窗口期（默认50）
    :param initial_capital: 初始资金（默认1000000）
    :param unit_size: 每次交易的交易量（默认100）
    :param risk_per_trade: 每次交易的最大风险占总资金的比例（默认0.01，表示1%的风险）
    :param atr_window: ATR指标的窗口期（默认14）
    :return: 总收益和平均每次交易收益
    """


    # 计算短期和长期移动平均线
    df['MA20'] = df['Close'].rolling(window=ma_short_window, min_periods=1).mean()
    df['MA50'] = df['Close'].rolling(window=ma_long_window, min_periods=1).mean()
    
    # 计算ATR（平均真实波幅）
    df['High-Low'] = df['High'] - df['Low']
    df['High-Close'] = abs(df['High'] - df['Close'].shift(1))
    df['Low-Close'] = abs(df['Low'] - df['Close'].shift(1))
    df['TR'] = df[['High-Low', 'High-Close', 'Low-Close']].max(axis=1)
    df['ATR'] = df['TR'].rolling(window=atr_window).mean()
    
    # 移除NaN值
    df.dropna(subset=['MA50', 'ATR'], inplace=True)
    
    # 确定买入和卖出信号
    df['Buy_Signal'] = df['Close'].gt(df['MA20']) & df['Close'].shift(1).lt(df['MA20'].shift(1))
    df['Sell_Signal'] = df['Close'].lt(df['MA20']) & df['Close'].shift(1).gt(df['MA20'].shift(1))

    # 计算每次交易的风险
    capital_per_trade = initial_capital * risk_per_trade  # 每次交易的最大风险金额
    df['Position'] = 0
    df.loc[df['Buy_Signal'], 'Position'] = unit_size
    df.loc[df['Sell_Signal'], 'Position'] = -unit_size
    
    # 根据 ATR 动态调整单位仓位
    # 假设每次交易的最大亏损是仓位乘以ATR


    # 止损和止盈策略（基于ATR）
    df['Stop_Loss'] = df['Close'] - 2 * df['ATR']
    df['Take_Profit'] = df['Close'] + 3 * df['ATR']

    
    # 计算最大仓位
    df['Adjusted_Unit_Size'] = capital_per_trade / (df['ATR'] * df['Close'])  # 使用 ATR 调整仓位大小
    df['Adjusted_Unit_Size'] = df['Adjusted_Unit_Size'].apply(lambda x: min(x, unit_size))  # 保证不会超过最大仓位
    
    # 计算持仓量和资金曲线
    df['Total_Value'] = df['Adjusted_Unit_Size'] * df['Close'].shift(-1)

    # 计算每日盈亏和总盈亏
    df['Daily_Return'] = df['Total_Value'].pct_change()
    # 清除NaN和inf值
    df['Daily_Return'].replace([np.inf, -np.inf], np.nan, inplace=True)
    df['Daily_Return'].fillna(0, inplace=True)

    df['Cumulative_Return'] = (df['Daily_Return'] + 1).cumprod()

    # 计算总收益和平均收益
    cumulative_returns = df['Cumulative_Return'].iloc[-1] * initial_capital - initial_capital
    total_trades = df[df['Position'] != 0].shape[0]
    average_return = cumulative_returns / total_trades if total_trades != 0 else 0

    return cumulative_returns, average_return

```

### 10.3.1 步骤1：数据获取和准备数据

#### 1. 处理空行和编码问题

```python
inputfile = 'data/0601857股票历史交易数据.csv'
# 目标文件
outfile = 'data/0601857股票历史交易数据（清洗后）.csv'
try:
    # 正确指定编码为 gbk
    with open(inputfile, 'r', encoding='gbk') as input_file, open(outfile, 'w', encoding='utf-8') as output_file:
        # 逐行读取原始文件
        for line in input_file:
            # 去除行末的换行符
            line = line.rstrip('\n')
            # 判断是否为空行
            if line:
                # 写入非空行到目标文件
                output_file.write(line + '\n')
    print('处理完成。')
except UnicodeDecodeError:
    print("GBK 编码尝试失败，请尝试其他编码。")
```

#### 2. 读取数据

```python
import pandas as pd
# 数据文件
f =  'data/0601857股票历史交易数据（清洗后）.csv'
# 读取股票历史交易数据
df = pd.read_csv(f, index_col='日期', parse_dates=True)
df # 打印df 
```

#### 3. 处理数据冗余

```python
# 移除“股票代码”和“名称”列
df = df.drop(['股票代码','名称'], axis=1)
df # 打印df 
```

#### 4. 过滤数据

```python
# 将日期列转换为日期时间类型
# 筛选出2021年的数据
# df = df.query('日期.dt.year == 2021')
# # 打印前10条数据
# df.head(10)	
```

#### 5. 处理中文列名

```python
# 重新命名列名
column_mapping = {	
    '日期': 'Date',
    '收盘价': 'Close',
    '最高价': 'High',
    '最低价': 'Low',
    '开盘价': 'Open',
    '成交量':'Volume'
}
df = df.rename(columns=column_mapping)	
# # 打印前10条数据
# df.head(10)
```

### 10.3.3 步骤3：定义海龟交易策略规则

```python
# 示例调用
ma_short_window = 20
ma_long_window = 50
initial_capital = 1000000
unit_size = 100
risk_per_trade = 0.01  # 每次交易承担的风险为1%

# 创建一个df副本并传递给函数
# df_copy = df.copy()

# 调用策略函数
total_profit, average_return = turtle_trading_strategy(df, ma_short_window=ma_short_window, ma_long_window=ma_long_window, initial_capital=initial_capital, unit_size=unit_size, risk_per_trade=risk_per_trade)

print(f"总交易次数：{df[df['Position'] != 0]['Position'].count()}")
print(f"总盈利：{total_profit:.2f}元")
print(f"平均收益：{average_return:.2f}元/交易")
```
