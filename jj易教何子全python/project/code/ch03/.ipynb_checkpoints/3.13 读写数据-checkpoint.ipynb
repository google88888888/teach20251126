### 3.13.4 	示例：写入银行账户交易记录到CSV文件

```python
import pandas as pd

# 示例银行账户交易记录
data = {
    "交易日期": ["2025-02-01", "2025-02-02", "2025-02-03", "2025-02-04", "2025-02-05"],
    "交易类型": ["存款", "取款", "存款", "取款", "存款"],
    "交易金额": [5000, 2000, 3000, 1000, 1500],
    "账户余额": [10000, 8000, 11000, 10000, 11500]
}

# 创建 DataFrame
df = pd.DataFrame(data)

# 写入到 CSV 文件
df.to_csv('data/account_transactions.csv', index=False, encoding='utf-8')

print("银行账户交易记录已成功写入 CSV 文件")
```

### 3.13.8 	案例：从数据库读取行账户交易记录数据

```python
import sqlite3
import pandas as pd

# 连接到 SQLite 数据库
conn = sqlite3.connect('data/bank_db.db')  # 替换为你的数据库文件路径

# 使用 pandas 从数据库读取数据
query = "SELECT * FROM account_transactions;"
df = pd.read_sql(query, conn)

# 关闭连接
conn.close()

# 打印读取的数据
df
```
